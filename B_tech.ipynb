```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==============================
# 1. File paths
# ==============================
csv_path = "/content/perceptual-image-hashing-cnn/results/threshold_metrics.csv"
results_dir = "/content/perceptual-image-hashing-cnn/results"
output_path = os.path.join(results_dir, "confusion_matrix.png")

# Ensure results folder exists
os.makedirs(results_dir, exist_ok=True)

# ==============================
# 2. Load threshold metrics
# ==============================
df = pd.read_csv(csv_path)

# Check required columns
required_columns = ["TP", "FP", "FN", "TN", "TPR", "FPR", "epsilon"]
missing_columns = [col for col in required_columns if col not in df.columns]

if missing_columns:
    raise ValueError(f"Missing required columns in CSV: {missing_columns}")

# ==============================
# 3. Find best threshold using Youden's J
# ==============================
df["Youden_J"] = df["TPR"] - df["FPR"]

best_row = df.loc[df["Youden_J"].idxmax()]

threshold = float(best_row["epsilon"])

# ==============================
# 4. Extract confusion matrix values
# ==============================
TP = int(best_row["TP"])
FP = int(best_row["FP"])
FN = int(best_row["FN"])
TN = int(best_row["TN"])

# Rows = Actual labels
# Columns = Predicted labels
conf_matrix = [
    [TP, FN],
    [FP, TN]
]

# ==============================
# 5. Plot confusion matrix
# ==============================
sns.set_theme(style="white", font_scale=1.1)

fig, ax = plt.subplots(figsize=(8, 6))

sns.heatmap(
    conf_matrix,
    annot=True,
    fmt=",d",
    cmap="Blues",
    linewidths=1.2,
    linecolor="white",
    xticklabels=["Similar", "Dissimilar"],
    yticklabels=["Similar", "Dissimilar"],
    annot_kws={"size": 15, "weight": "bold"},
    cbar=True,
    square=True,
    ax=ax
)

ax.set_xlabel("Predicted Label", fontsize=13, fontweight="bold", labelpad=10)
ax.set_ylabel("Actual Label", fontsize=13, fontweight="bold", labelpad=10)

ax.set_title(
    f"Confusion Matrix (Youden's J Threshold = {threshold:.4f})",
    fontsize=15,
    fontweight="bold",
    pad=16
)

ax.tick_params(axis="both", labelsize=11, length=0)

plt.tight_layout()

# ==============================
# 6. Save graph in results folder
# ==============================
fig.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight",
    facecolor="white"
)

plt.show()

# ==============================
# 7. Print results
# ==============================
print("Best Threshold (Youden's J):", round(threshold, 6))
print(f"TP: {TP:,}")
print(f"FP: {FP:,}")
print(f"FN: {FN:,}")
print(f"TN: {TN:,}")
print("\nConfusion matrix saved successfully:")
print(output_path)
```

```python
import os
import zipfile

repo_path = "/content/perceptual-image-hashing-cnn"
zip_path = os.path.join(repo_path, "results.zip")
png_path = os.path.join(repo_path, "results", "confusion_matrix.png")
temp_zip_path = zip_path + ".tmp"

# Check that both files exist
if not os.path.exists(zip_path):
    raise FileNotFoundError(f"ZIP not found: {zip_path}")

if not os.path.exists(png_path):
    raise FileNotFoundError(f"Confusion matrix not found: {png_path}")

# Rebuild ZIP, keeping existing contents and adding/updating the PNG
with zipfile.ZipFile(zip_path, "r") as old_zip:
    with zipfile.ZipFile(temp_zip_path, "w", zipfile.ZIP_DEFLATED) as new_zip:
        for item in old_zip.infolist():
            if item.filename != "confusion_matrix.png":
                new_zip.writestr(item, old_zip.read(item.filename))

        new_zip.write(png_path, "confusion_matrix.png")

os.replace(temp_zip_path, zip_path)

print("Updated results.zip successfully!")
```
